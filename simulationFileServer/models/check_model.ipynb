```python
import torch

model_path = 'sac_model_3.pth'
checkpoint = torch.load(model_path, map_location='cpu')

print("🧾 Các khóa (keys) có trong file:")
for key in checkpoint.keys():
    print(f"  - {key}")

print("\n📐 Kích thước của từng phần:")
for key, value in checkpoint.items():
    if isinstance(value, dict):
        print(f"{key}:")
        for param_name, param_tensor in value.items():
            if hasattr(param_tensor, 'shape'):
                print(f"    {param_name}: {tuple(param_tensor.shape)}")
            else:
                print(f"    {param_name}: (not a tensor) -> type={type(param_tensor)}")
    else:
        print(f"{key}: {value}")
```
