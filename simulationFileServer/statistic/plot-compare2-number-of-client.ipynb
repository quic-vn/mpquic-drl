```python
import pandas as pd
import matplotlib.pyplot as plt

var = 0
los = 0
scenario = 'none'
column_names = ['Download Time']

# Danh sách các scheduler
schedulers = ['rtt', 'fuzzyqsat', 'sac']
labels = ['minRTT', 'FQ-SAT', 'SAC-based']

# Số lượng client
num_clients = 4

# Mảng lưu trữ dữ liệu
download_times = {scheduler: [] for scheduler in schedulers}

# Load dữ liệu cho từng scheduler và từng client
for scheduler in schedulers:
    for i in range(1, num_clients + 1):
        file_name = f'../output/{scenario}-{num_clients}-2MB-0-0-50-10-{var}-{los}-{scheduler}-result{i+2}.csv'
        df = pd.read_csv(file_name, names=column_names)
        download_times[scheduler].append(df['Download Time'] / 1000)  # Chuyển đổi sang giây và lưu vào mảng

# Tạo boxplot
plt.figure(figsize=(6, 6))
plt.rcParams.update({'font.size': 16, 'font.family': 'sans-serif'})

# Vị trí cho từng boxplot
width = 0.15  # Độ rộng của mỗi box
gap = 0.08  # Khoảng cách giữa các box trong cùng nhóm
group_gap = 0.3  # Khoảng cách giữa các nhóm client
positions = []
base_positions = range(1, num_clients + 1)

for i, scheduler in enumerate(schedulers):
    positions += [p + i * (width + gap) + group_gap * (p - 1) for p in base_positions]

# Tạo boxplot cho từng scheduler
colors = ['lightblue', 'lightgreen', 'violet', 'purple']
for i, scheduler in enumerate(schedulers):
    box = plt.boxplot(download_times[scheduler],
                      positions=[p + i * (width + gap) + group_gap * (p - 1) for p in base_positions],
                      widths=width, patch_artist=True,
                      flierprops=dict(marker='o', color='red', markersize=5), showfliers=False)
    for patch in box['boxes']:
        patch.set_facecolor(colors[i])

# Thiết lập nhãn và legend
plt.xticks([p + (len(schedulers) - 1) * (width + gap) / 2 + group_gap * (p - 1) for p in base_positions],
           [str(i) for i in base_positions])
plt.xlabel('Client')
plt.ylabel('Download Time (s)')
plt.title(scenario)
plt.grid(True)

# Thêm legend
handles = [plt.Line2D([0], [0], color=color, lw=4) for color in colors]
plt.legend(handles, labels, loc='upper left')

plt.ylim(0.4, 2)
plt.show()
```
