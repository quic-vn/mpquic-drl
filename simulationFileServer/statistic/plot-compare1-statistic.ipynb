```python
import pandas as pd
import matplotlib.pyplot as plt

column_names = ['Download Time']
schedules = ['rtt', 'fuzzyqsat', 'sac']
var = 10
los = 1

# List to hold all download time data
all_download_times = {sch: [] for sch in schedules}

# Common scenarios
scenarios = [
    "5-10", "5-20", "10-10", "10-20", "30-10", "30-20"
]

for sch in schedules:
    for scenario in scenarios:
        # Load the data from the CSV file
        data = pd.read_csv(f'../output/aug2/compare-1/none-1-2MB-0-0-{scenario}-{var}-{los}-{sch}-result3.csv', names=column_names)
        
        # Extract and store the Download Time column
        all_download_times[sch].append(data['Download Time'] / 1000)

# Create a boxplot
plt.figure(figsize=(8, 6))
plt.rcParams.update({'font.size': 14, 'font.family': 'sans-serif'})

# Prepare data and positions for the boxplot
positions = []
base_position = 1

colors = {'rtt': 'lightblue', 'fuzzyqsat': 'lightgreen', 'sac': 'violet'}
boxplots = []

for i, sch in enumerate(schedules):
    box = plt.boxplot([all_download_times[sch][j] for j in range(len(scenarios))], 
                      positions=[1 + x + i * 0.3 for x in range(len(scenarios))],
                      widths=0.25, patch_artist=True, 
                      flierprops=dict(marker='o' if sch == 'rtt' else 'x', markersize=5), showfliers=False)
    for patch in box['boxes']:
        patch.set_facecolor(colors[sch])
    boxplots.append(box)

# Set x-ticks and labels
plt.xticks([i + 0.3 for i in range(1, len(scenarios) + 1)], scenarios)
plt.xlabel('Scenarios [BW (Mbps), One-way delay (ms)]')
plt.ylabel('Download Time (s)')
plt.grid(True)

# Add legend
plt.legend([box["boxes"][0] for box in boxplots], ['minRTT-based', 'FQ-SAT', 'SAC-based'], loc='upper left')
plt.ylim(0.4, 2.3)

plt.show()
```
