```python
# Tính giá trị trung bình và độ lệch chuẩn cho mỗi kịch bản
scenarios = [1, 4, 8]

def calculate_mean_std(data_1, data_4, data_8, column):
    mean = [data_1[column].mean(), data_4[column].mean(), data_8[column].mean()]
    std = [data_1[column].std(), data_4[column].std(), data_8[column].std()]
    return mean, std

rtt_cwnd_mean_1, rtt_cwnd_std_1 = calculate_mean_std(rtt_1, rtt_4, rtt_8, 'CWND_1')
rtt_cwnd_mean_2, rtt_cwnd_std_2 = calculate_mean_std(rtt_1, rtt_4, rtt_8, 'CWND_2')
fuzzyqsat_cwnd_mean_1, fuzzyqsat_cwnd_std_1 = calculate_mean_std(fuzzyqsat_1, fuzzyqsat_4, fuzzyqsat_8, 'CWND_1')
fuzzyqsat_cwnd_mean_2, fuzzyqsat_cwnd_std_2 = calculate_mean_std(fuzzyqsat_1, fuzzyqsat_4, fuzzyqsat_8, 'CWND_2')

rtt_srtt_mean_1, rtt_srtt_std_1 = calculate_mean_std(rtt_1, rtt_4, rtt_8, 'sRTT_1')
rtt_srtt_mean_2, rtt_srtt_std_2 = calculate_mean_std(rtt_1, rtt_4, rtt_8, 'sRTT_2')
fuzzyqsat_srtt_mean_1, fuzzyqsat_srtt_std_1 = calculate_mean_std(fuzzyqsat_1, fuzzyqsat_4, fuzzyqsat_8, 'sRTT_1')
fuzzyqsat_srtt_mean_2, fuzzyqsat_srtt_std_2 = calculate_mean_std(fuzzyqsat_1, fuzzyqsat_4, fuzzyqsat_8, 'sRTT_2')

rtt_restran_mean_1, rtt_restran_std_1 = calculate_mean_std(rtt_1, rtt_4, rtt_8, 'RestranPacket_1')
rtt_restran_mean_2, rtt_restran_std_2 = calculate_mean_std(rtt_1, rtt_4, rtt_8, 'RestranPacket_2')
fuzzyqsat_restran_mean_1, fuzzyqsat_restran_std_1 = calculate_mean_std(fuzzyqsat_1, fuzzyqsat_4, fuzzyqsat_8, 'RestranPacket_1')
fuzzyqsat_restran_mean_2, fuzzyqsat_restran_std_2 = calculate_mean_std(fuzzyqsat_1, fuzzyqsat_4, fuzzyqsat_8, 'RestranPacket_2')

# Vẽ biểu đồ với error bars
fig, axs = plt.subplots(1, 3, figsize=(15, 5))

# Biểu đồ 1: CWND
line1 = axs[0].errorbar(scenarios, rtt_cwnd_mean_1, yerr=rtt_cwnd_std_1, color='blue', linestyle='-', marker='o', label='RTT-1')
line2 = axs[0].errorbar(scenarios, rtt_cwnd_mean_2, yerr=rtt_cwnd_std_2, color='blue', linestyle='--', marker='o', label='RTT-2')
line3 = axs[0].errorbar(scenarios, fuzzyqsat_cwnd_mean_1, yerr=fuzzyqsat_cwnd_std_1, color='green', linestyle='-', marker='o', label='FuzzyQSAT-1')
line4 = axs[0].errorbar(scenarios, fuzzyqsat_cwnd_mean_2, yerr=fuzzyqsat_cwnd_std_2, color='green', linestyle='--', marker='o', label='FuzzyQSAT-2')
axs[0].set_title('Average CWND')
axs[0].set_xlabel('Scenario')
axs[0].set_ylabel('Average CWND')
axs[0].set_xticks(scenarios)

# Biểu đồ 2: sRTT
axs[1].errorbar(scenarios, rtt_srtt_mean_1, yerr=rtt_srtt_std_1, color='blue', linestyle='-', marker='o')
axs[1].errorbar(scenarios, rtt_srtt_mean_2, yerr=rtt_srtt_std_2, color='blue', linestyle='--', marker='o')
axs[1].errorbar(scenarios, fuzzyqsat_srtt_mean_1, yerr=fuzzyqsat_srtt_std_1, color='green', linestyle='-', marker='o')
axs[1].errorbar(scenarios, fuzzyqsat_srtt_mean_2, yerr=fuzzyqsat_srtt_std_2, color='green', linestyle='--', marker='o')
axs[1].set_title('Average sRTT')
axs[1].set_xlabel('Scenario')
axs[1].set_ylabel('Average sRTT')
axs[1].set_xticks(scenarios)

# Biểu đồ 3: RestranPacket
axs[2].errorbar(scenarios, rtt_restran_mean_1, yerr=rtt_restran_std_1, color='blue', linestyle='-', marker='o')
axs[2].errorbar(scenarios, rtt_restran_mean_2, yerr=rtt_restran_std_2, color='blue', linestyle='--', marker='o')
axs[2].errorbar(scenarios, fuzzyqsat_restran_mean_1, yerr=fuzzyqsat_restran_std_1, color='green', linestyle='-', marker='o')
axs[2].errorbar(scenarios, fuzzyqsat_restran_mean_2, yerr=fuzzyqsat_restran_std_2, color='green', linestyle='--', marker='o')
axs[2].set_title('Average RestranPacket (%)')
axs[2].set_xlabel('Scenario')
axs[2].set_ylabel('Average RestranPacket (%)')
axs[2].set_xticks(scenarios)

# Thêm legend chung
fig.legend([line1, line2, line3, line4], ['RTT-1', 'RTT-2', 'FuzzyQSAT-1', 'FuzzyQSAT-2'], loc='upper center', ncol=4)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
