```python
import pandas as pd
import matplotlib.pyplot as plt

# Các thuật toán cần so sánh
schedules = ['rtt', 'fuzzyqsat', 'sac']
var = 10
los = 0.5

# Danh sách các kịch bản
scenarios = [
    "none"
]

# Dictionary để lưu dữ liệu thông lượng cho từng thuật toán
all_throughput_times = {sch: [] for sch in schedules}

# Màu sắc cho từng thuật toán
colors = {'rtt': 'blue', 'fuzzyqsat': 'green', 'sac': 'violet'}
labels = {'rtt': 'minRTT', 'fuzzyqsat': 'FQ-SAT', 'sac': 'SAC-based'}
line_styles = {'rtt': '--', 'fuzzyqsat': '-.', 'sac': '-'}

# Đọc dữ liệu và tính thông lượng
for sch in schedules:
    for scenario in scenarios:
        # Đọc dữ liệu từ file CSV tương ứng
        data = pd.read_csv(f'../output/{scenario}-1-500MB-0-0-25-5-{var}-{los}-{sch}-result3.csv')
        # Tính thông lượng (khác biệt giữa các giá trị liên tiếp và đổi sang MB/s)
        all_throughput_times[sch] = data.diff().fillna(0) / 1e6

# Vẽ biểu đồ thông lượng theo thời gian
plt.figure(figsize=(10, 6))
plt.rcParams.update({'font.size': 16, 'font.family': 'sans-serif'})

for sch in schedules:
    plt.plot(all_throughput_times[sch], label=labels[sch], color=colors[sch], linestyle=line_styles[sch])

# Thêm tiêu đề và nhãn trục
# plt.title('Throughput Over Time for RTT, FuzzyQSAT, and SAC (MB/s)')
plt.xlabel('Time (s)')
plt.ylabel('Throughput (MB/s)')
plt.legend()
plt.xlim(0, 60)
plt.ylim(6, 8)
plt.grid(True)

# Hiển thị biểu đồ
plt.show()
```
